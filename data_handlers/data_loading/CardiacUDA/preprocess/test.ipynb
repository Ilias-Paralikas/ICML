```python
from torch.utils.data import ConcatDataset
from original_dataset import SiteDataset
import os
import numpy as np
from scipy.ndimage import center_of_mass
import matplotlib.pyplot as plt


root = '../../../data/CardiacUDA/original_data/train'
sites = os.listdir(root)

datasets =[]
for site in sites:
    site_path = os.path.join(root,site)
    dataset = SiteDataset(site_path)
    datasets.append(dataset)

dataset = ConcatDataset(datasets)
```

```python

shape_buckets ={}
for j in range(len(datasets)):
    d = datasets[j]
    for i in range(len(d)):
        if d[i][0].shape[:2] not in shape_buckets:
            shape_buckets[d[i][0].shape[:2]] = [(j,i)]
        else:
            shape_buckets[d[i][0].shape[:2]].append((j,i))
```
