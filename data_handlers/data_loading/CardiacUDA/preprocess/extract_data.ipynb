```python
import os
import shutil

root_train_folder ='cardiacUDC_dataset/train'
os.makedirs(root_train_folder, exist_ok=True)
for folder in os.listdir('cardiacUDC_dataset'):
    folder_path = os.path.join('cardiacUDC_dataset', folder)
    if os.path.isdir(folder_path) and folder.startswith('Site_'):
        shutil.move(folder_path, os.path.join(root_train_folder, folder))
```
