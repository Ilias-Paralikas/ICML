```python
i = i+1
x = dataset[i][0]
print(i)
centers_2d =[]
slices =x.shape[2]
for s in range(slices):
    c =center_of_mass(x[:,:,s])
    centers_2d.append(c)

centers_2d = np.array(centers_2d)


# Separate x and y coordinates
x_coords = [c[0] for c in centers_2d]
y_coords = [c[1] for c in centers_2d]

plt.figure(figsize=(8, 8))
plt.scatter(x_coords, y_coords, c='red', s=100, marker='o')  # s controls size
plt.grid(True)
plt.gca().invert_yaxis()  # match array coordinates
plt.grid(False)
plt.show()
```

```python
weirds = []
for j,c in enumerate(centers_2d):
    if c[0]>460:
        weirds.append(j)
len(weirds)

x,_,path= dataset[i]
plt.imshow(x[:,:,weirds[0]])
```
